```python

import sqlite3
import pandas as pd

con = sqlite3.connect("lab2_customers.db")
```

```python

create_table = """
CREATE TABLE customers (
    id INTEGER PRIMARY KEY,
    name TEXT,
    city TEXT,
    age INTEGER
)
"""

con.execute(create_table)
```

```python

insert_customers = """
INSERT INTO customers (id, name, city, age)
VALUES
    (1, 'Alex Rivera', 'Los Angeles', 25),
    (2, 'Jordan Lee', 'Santa Monica', 31),
    (3, 'Taylor Smith', 'Inglewood', 28),
    (4, 'Morgan Davis', 'Long Beach', 35),
    (5, 'Casey Brown', 'Torrance', 22),
    (6, 'Riley Wilson', 'Culver City', 29),
    (7, 'Jamie Garcia', 'Pasadena', 40),
    (8, 'Avery Martin', 'Burbank', 33)
"""

con.execute(insert_customers)
con.commit()
```

```python

result = pd.read_sql("SELECT * FROM customers", con)
result
```

I created a customers table using CREATE TABLE and added eight fictional customer records using INSERT. I used SELECT to verify that all eight rows were successfully added to the database.
