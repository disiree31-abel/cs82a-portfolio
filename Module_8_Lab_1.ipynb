```python
import pandas as pd

df = pd.read_csv("penguins.csv")

# Display the first five rows
display(df.head())
```

```python
# 1. Dataset shape
print("Dataset shape:", df.shape)

# 2. Missing values in each column
print("\nMissing values:")
print(df.isnull().sum())

# 3. Number of penguins in each species
print("\nPenguins per species:")
print(df["species"].value_counts())

# 4. Mean flipper length per species
print("\nMean flipper length by species:")
print(df.groupby("species")["flipper_length_mm"].mean())
```
