```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_clean.csv")

print(df.shape)
df.head()
```

```python
print("PRICE")
print("Mean:", df["price"].mean())
print("Median:", df["price"].median())
print("Standard deviation:", df["price"].std())

print("\nQTY")
print("Mean:", df["qty"].mean())
print("Median:", df["qty"].median())
print("Standard deviation:", df["qty"].std())
```

Step 3: Mean or Median?

I would report the median price because the mean is noticeably higher than the median. This difference suggests that the price data is right skewed, with some expensive items pulling the mean upward. The median better represents the typical price in this situation.

```python
df["price"].plot(kind="hist", bins=20)

plt.title("Distribution of Price")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

Step 4: Histogram Shape

The price distribution is right skewed, with most prices concentrated at the lower end and a tail extending toward higher prices.

```python
df["revenue"] = df["price"] * df["qty"]

df.head()
```

```python
correlation_matrix = df.corr(numeric_only=True)

correlation_matrix
```

```python
df.plot.scatter(x="qty", y="revenue")

plt.title("Quantity vs Revenue")
plt.xlabel("Quantity")
plt.ylabel("Revenue")
plt.show()
```

Step 8: Analyst's Sentence

There is a strong positive relationship between quantity sold and revenue (r = 0.85), meaning that higher quantities tend to be associated with higher revenue; however, price differences and unusually high value transactions could also affect revenue, so this correlation does not by itself show that increasing quantity causes higher revenue.
