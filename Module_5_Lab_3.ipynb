```python
import pandas as pd
```

```python
cust = pd.read_csv("custdata (1).tsv", sep="\t")
cust.head()
```

```python
cust.columns
```

```python
cust[["age", "income", "num.vehicles"]].describe()
```

```python
before_cleaning = cust[["age", "income", "num.vehicles"]].corr()

before_cleaning
```

```python
print("Negative incomes:", (cust["income"] < 0).sum())
print("Impossible ages:", ((cust["age"] < 0) | (cust["age"] > 120)).sum())
```

```python
cust_clean = cust[
    (cust["age"] >= 0) &
    (cust["age"] <= 120) &
    (cust["income"] >= 0)
]
```

```python
print("Original rows:", len(cust))
print("Cleaned rows:", len(cust_clean))
print("Rows removed:", len(cust) - len(cust_clean))
```

```python
after_cleaning = cust_clean[["age", "income", "num.vehicles"]].corr()

after_cleaning
```
