```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
cust = pd.read_csv("custdata.tsv", sep="\t")
cust.head()
```

```python
cust.columns
```

```python
cust["housing.type"].value_counts(dropna=False)
```

```python
before = len(cust)

cust_clean = cust.dropna(subset=["housing.type"])

after = len(cust_clean)

print("Rows excluded:", before - after)
```

```python
housing_counts = cust_clean["housing.type"].value_counts().sort_values()

housing_counts
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=housing_counts.index,
    y=housing_counts.values
)

plt.title("Number of Customers by Housing Type")
plt.xlabel("Housing Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

I excluded 56 rows because they had missing housing information.
