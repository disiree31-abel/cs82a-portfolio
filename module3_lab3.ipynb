Plausibility Assumption

I am assuming that a bridge length between 500 and 4,000 feet is a reasonable range for this dataset. I will flag anything outside this range so I can take a closer look at it. I chose this range because most of the lengths in the dataset are between about 800 and 2,000 feet, so a length over 4,000 feet stands out from the rest.

```python
flagged = df[(df["length"] < 500) | (df["length"] > 4000)]

display(flagged)
```

```python
print("Number of flagged rows:", len(flagged))
```

```python
print("Flagged lengths:")
print(flagged["length"].tolist())
```

```python
display(flagged)
```
