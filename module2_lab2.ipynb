```python
def multiply_for(a, b):
    result = 0

    for i in range(b):
        result = result + a

    return result
```

```python
print("7 x 6 =", multiply_for(7, 6))
```

```python
def multiply_while(a, b):
    result = 0
    count = 0

    while count < b:
        result = result + a
        count = count + 1

    return result
```

```python
print("7 x 6 =", multiply_while(7, 6))
```

```python
a = 7
b = 6

for_result = multiply_for(a, b)
while_result = multiply_while(a, b)

print("For loop result:", for_result)
print("While loop result:", while_result)
print("Both agree:", for_result == while_result)
```

```python
real_result = 7 * 6

print("For loop:", for_result)
print("While loop:", while_result)
print("Real product:", real_result)
```

```python
print("5 x 4 =", multiply_for(5, 4))
print("5 x 4 =", multiply_while(5, 4))
```
